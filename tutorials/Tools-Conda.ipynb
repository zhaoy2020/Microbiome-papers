# 3. <a id='toc3_'></a>[基本操作](#toc0_)

```bash
# create
conda create -n deepLearning python tensonflow -y                       # 创建环境并安装python tensonflow包
conda create -n deepLearning --file requrienments.txt                   # 按照requrienments.txt软件列表创建环境并安装软件

# rename
conda rename -n deepLearning deepLearning1

# remove
conda env remove -n deepLearning -y                                     # 删除环境
conda remove -n deepLearning --all -y                                   # 删除环境

# 移植程序1
conda list -e > requirements.txt
conda install --yes --file requirements.txt

# 移植程序2 （推荐）
conda env export > environment.yml                           # export
conda env create -f environment.yml                          # install from environment.yml --file 

conda env list                                                          # list所有虚拟环境
conda activate deepLearning                                             # 激活环境
conda deactivate      
```

```bash
conda search python                                                     # 搜索包
conda install python=3.6 -y                                             # 安装指定版本包
conda install --file requriment.txt -n deepLearning -p /home/xxx/xxx    # 依照requriment.txt并在指定环境中和指定位置上安装包
conda remove python -y                                                  # 卸载包
conda update python -y                                                  # 升级包
conda list    
```
