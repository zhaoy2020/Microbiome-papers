# 1. <a id='toc1_'></a>[About Jupyter](#toc0_)

![Jupyter](https://jupyter.org/assets/homepage/main-logo.svg) Jupyter 是一个开源的交互式计算环境，允许用户创建和共享包含代码、文本、公式、可视化图表和解释性内容的文档。它的名字来源于三种主要支持的编程语言：Julia、Python 和 R（Ju-Py-R），但如今 Jupyter 已经支持超过 40 种编程语言。

主要特点：
- 交互式笔记本（Notebook）
  - Jupyter 最著名的组件是 Jupyter Notebook，它允许你在一个网页界面中编写和运行代码，每段代码在一个“单元格”中执行，并即时显示结果。这种交互方式非常适合数据探索、教学和演示。
- 支持多种语言（内核）
  - 虽然最常用于 Python（通过 IPython 内核），但 **Jupyter** 支持多种语言，如 **R**、Julia、Scala、**C++** 等，只需安装相应的“内核”（kernel）。
- 富文本支持（Markdown）
  - 除了代码，你还可以使用 Markdown 编写说明文字、插入图片、数学公式（LaTeX）、链接等，使得文档兼具代码和文档功能，非常适合做报告或教学笔记。
- 可视化集成
  -可以直接在笔记本中生成图表（如使用 Matplotlib、Seaborn、Plotly 等），结果会内嵌显示。
- 易于分享和发布
  - Jupyter Notebook 文件（.ipynb）可以导出为 **HTML**、**PDF**、Slides 等格式，也支持在 GitHub、Jupyter Book 或 nbviewer 上展示。
- 广泛应用领域
  - 常用于数据科学、机器学习、数据分析、科学计算、教育和研究等领域。

# 2. <a id='toc2_'></a>[Jupyter in VScode](#toc0_)

Visual Studio Code（VS Code）中的 Jupyter 支持是近年来数据科学和 Python 开发领域的一项重要进步。通过微软官方提供的 Jupyter 扩展，VS Code 实现了对 `.ipynb` 笔记本文件的原生支持，使用户可以在熟悉的集成开发环境（IDE）中直接打开、编辑和运行 Jupyter Notebook。这一功能不仅保留了 Jupyter 交互式编程的核心优势——如分单元执行代码、即时查看输出结果（包括文本、图表和 Markdown 文档），还深度融合了现代 IDE 的强大特性，例如智能代码补全、语法检查、Git 版本控制、主题定制以及多文件项目管理。用户可以在 `.py` 脚本中使用 `# %%` 标记将代码划分为多个可独立运行的单元，实现“类 notebook”的交互式开发体验，同时也能直接编辑 `.ipynb` 文件，享受与传统 Jupyter 类似的界面布局。

相比之下，传统的 Jupyter 环境通常指通过浏览器访问的 Jupyter Notebook 或其现代化版本 JupyterLab。Jupyter Notebook 以简洁直观著称，适合快速实验和教学演示，而 JupyterLab 则提供了更接近 IDE 的多面板工作区，支持文件浏览器、终端、Markdown 编辑器等组件，适用于更复杂的分析任务。然而，这些基于浏览器的环境在代码编辑能力、调试支持和工程化管理方面存在局限。例如，它们缺乏原生的断点调试功能，代码补全和错误提示较弱，且在处理大型项目或多文件协作时不如 VS Code 高效。此外，`.ipynb` 文件包含代码、输出和元数据，容易导致 Git 版本冲突，而 VS Code 提供了更好的版本控制集成和变量查看器，帮助开发者更清晰地管理状态和调试逻辑。

从使用体验来看，VS Code 中的 Jupyter 更适合追求高效、可维护和可调试工作流的专业用户。它特别适用于机器学习建模、数据分析脚本开发以及需要将探索性分析转化为生产代码的场景。开发者可以在同一个环境中完成从数据探索到模块化封装的全过程，无需在不同工具之间切换。而传统 Jupyter 环境则在教学、报告撰写和快速原型设计方面更具优势，因其输出结果可直接导出为 HTML 或 PDF，并可通过 nbviewer 等平台轻松分享，具有更强的展示性和可读性。对于初学者而言，Jupyter Notebook 的低门槛和即时反馈机制有助于快速理解编程概念和数据处理流程。
